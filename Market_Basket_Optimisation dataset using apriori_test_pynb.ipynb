```python
transcation=[]
for i in range (0,len(df)):
  transcation.append([str(df.values[i,j])
  for j in range(0,20)if str(df.values[i,j])!='nan'])
transcation
```

```python
from apyori import apriori
```

```python
rules=apriori(transactions=transcation,min_support=0.002,min_confidence=0.2,min_lift=3,min_length=2,max_lenght=2)
```

```python
results=list(rules)
results
```

```python
import pandas as pd

data = []

for r in results:
    support = r.support

    for stat in r.ordered_statistics:
        data.append([
            list(stat.items_base),
            list(stat.items_add),
            support,
            stat.confidence,
            stat.lift
        ])

df = pd.DataFrame(
    data,
    columns=["LHS", "RHS", "Support", "Confidence", "Lift"]
)

df
```

```python
df_sorted=df.sort_values(
    by=['Lift','Confidence'],
    ascending=[False,False]
)
df_sorted
```
