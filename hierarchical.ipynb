```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
df = pd.read_csv("Mall_Customers.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
X = df.iloc[:, [3, 4]].values
```

```python
plt.figure(figsize=(10,5))
dendrogram(linkage(X, method='ward'))
```

```python
from sklearn.cluster import AgglomerativeClustering
```

```python
hc = AgglomerativeClustering(
    n_clusters=5,
    metric='euclidean',
    linkage='ward'
)

y_hc = hc.fit_predict(X)
```

```python
plt.figure(figsize=(8,6))
```

```python
plt.scatter(X[y_hc == 0, 0], X[y_hc == 0, 1], s=100,color='green', label='Cluster 1')
plt.scatter(X[y_hc == 1, 0], X[y_hc == 1, 1], s=100, color='pink',label='Cluster 2')
plt.scatter(X[y_hc == 2, 0], X[y_hc == 2, 1], s=100, color='blue',label='Cluster 3')
plt.scatter(X[y_hc == 3, 0], X[y_hc == 3, 1], s=100,color='black', label='Cluster 4')
plt.scatter(X[y_hc == 4, 0], X[y_hc == 4, 1], s=100,color='red', label='Cluster 5')
```

```python
methods = ['single', 'complete', 'average', 'ward']
```

```python
for method in methods:
    plt.figure(figsize=(8,4))
    dendrogram(linkage(X, method=method))
    plt.title(f"Dendrogram ({method.capitalize()} Linkage)")
    plt.xlabel("Customers")
    plt.ylabel("Distance")
    plt.show()
```
